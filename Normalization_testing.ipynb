```python
%%bash
cat > /content/check_reinhard.py <<'PY'
import cv2
import numpy as np

def stats(path):
    image = cv2.imread(path)
    image = cv2.cvtColor(image, cv2.COLOR_BGR2LAB)
    image = image.astype(np.float32)

    return image.mean(axis=(0, 1)), image.std(axis=(0, 1))

for name, path in [
    ("Target", "/content/target.png"),
    ("Original source", "/content/source.png"),
    ("Reinhard output", "/content/reinhard_output.png")
]:
    mean, std = stats(path)
    print(f"\n{name}")
    print("LAB means:", np.round(mean, 2))
    print("LAB stds:", np.round(std, 2))
PY

conda run -n histology38 python /content/check_reinhard.py
```

```python
%%bash

cat > /content/check_reinhard_standardized.py <<'PY'
import cv2
import numpy as np
from src.normalization import stain_utils

def stats(path):
    image = cv2.imread(path)
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    image = stain_utils.standardize_brightness(image)

    lab = cv2.cvtColor(image, cv2.COLOR_RGB2LAB)
    lab = lab.astype(np.float32)

    return lab.mean(axis=(0, 1)), lab.std(axis=(0, 1))

for name, path in [
    ("Standardized target", "/content/target.png"),
    ("Standardized source", "/content/source.png"),
    ("Reinhard output", "/content/reinhard_output.png")
]:
    mean, std = stats(path)

    print(f"\n{name}")
    print("LAB means:", np.round(mean, 2))
    print("LAB stds:", np.round(std, 2))
PY

conda run --no-capture-output -n histology38 env PYTHONPATH=/content/StainNormalization python -u /content/check_reinhard_standardized.py
```

```python
%%bash

cat > /content/check_macenko.py <<'PY'
import cv2
import numpy as np
from src.normalization import stainNorm_Macenko, stain_utils

def load_rgb(path):
    image = cv2.imread(path)
    return cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

target = load_rgb("/content/target.png")
source = load_rgb("/content/source.png")
output = load_rgb("/content/macenko_output.png")

normalizer = stainNorm_Macenko.normalizer()
normalizer.fit(target)

# Estimate concentrations using the fitted target stain matrix
target_concentrations = stain_utils.get_concentrations(
    target, normalizer.stain_matrix_target
)

output_concentrations = stain_utils.get_concentrations(
    output, normalizer.stain_matrix_target
)

target_p99 = np.percentile(target_concentrations, 99, axis=0)
output_p99 = np.percentile(output_concentrations, 99, axis=0)

print("Target 99th percentile concentrations:",
      np.round(target_p99, 3))

print("Output 99th percentile concentrations:",
      np.round(output_p99, 3))

print("\nTarget stain matrix:")
print(normalizer.stain_matrix_target)

print("\nAre all output concentrations finite?",
      np.isfinite(output_concentrations).all())

print("Output concentration range:",
      np.round(output_concentrations.min(), 3),
      "to",
      np.round(output_concentrations.max(), 3))
PY

conda run --no-capture-output -n histology38 env PYTHONPATH=/content/StainNormalization python -u /content/check_macenko.py
```
